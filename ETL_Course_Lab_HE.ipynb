<p>התא הבא אינו מתקין דבר. הוא מזהה את הפרויקט ואת ה־Python שנבחר ב־Cursor, ובודק שהחבילות המתאימות נמצאות באותה סביבה. אם נבחר Kernel אחר, יש לתקן את הבחירה לפני ההמשך.</p>

```python
from pathlib import Path
from datetime import datetime, timezone
from contextlib import closing
from decimal import Decimal
import importlib.util
import importlib.metadata as metadata
import hashlib
import html
import json
import os
import platform
import re
import shutil
import sqlite3
import struct
import subprocess
import sys
import time
import uuid
from IPython.display import HTML, display

os.environ["HF_HUB_DISABLE_TELEMETRY"] = "1"

# Optional: set a path here only if automatic project discovery fails.
PROJECT_FOLDER = None  # Example on Windows: r"C:\projects\etl-course-demo"

search_from = (Path(PROJECT_FOLDER).expanduser() if PROJECT_FOLDER else Path.cwd()).resolve()
PROJECT_ROOT = next((p for p in [search_from, *search_from.parents]
                     if (p / "scripts/generate_data.py").is_file()
                     and (p / "requirements-core.txt").is_file()), None)
if PROJECT_ROOT is None:
    raise RuntimeError("לא נמצאה תיקיית הפרויקט. פתח את המחברת בתיקיית etl-course-demo או הגדר PROJECT_FOLDER.")
if sys.version_info[:2] not in {(3, 11), (3, 12), (3, 13)}:
    raise RuntimeError("גרסה זו נבדקת עם Python 3.13.12. בחר Kernel עם Python 3.11, 3.12 או 3.13 והפעל מחדש.")
if struct.calcsize("P") * 8 != 64:
    raise RuntimeError("נדרשת סביבת Python של 64 ביט.")

# Read the project's exact requirements, including the core requirements file.
required = {}
for requirement_file in ["requirements-core.txt", "requirements.txt"]:
    for line in (PROJECT_ROOT / requirement_file).read_text(encoding="utf-8").splitlines():
        match = re.match(r"^([A-Za-z0-9_.-]+)(?:\[[^]]+\])?==([^\s#]+)", line.strip())
        if match:
            required[match.group(1)] = match.group(2)

version_rows, problems = [], []
for package, expected in required.items():
    try:
        actual = metadata.version(package)
    except metadata.PackageNotFoundError:
        actual = "MISSING"
    matches = actual.split("+")[0] == expected
    version_rows.append({"package": package, "expected": expected, "installed": actual,
                         "status": "OK" if matches else "CHECK"})
    if not matches:
        problems.append(f"{package}: {actual}; expected {expected}")
if problems:
    raise RuntimeError("יש להתקין את דרישות הפרויקט באותו Kernel, לפי ההוראות בתחילת המחברת:\n" + "\n".join(problems))

import pandas as pd
import numpy as np
import duckdb

STYLE = """
.etl-text,.etl-block{font-family:Arial,sans-serif;line-height:1.65;text-align:right;direction:rtl}
.etl-block{margin:10px 0;padding:12px 16px;border:1px solid #8c99aa;border-radius:8px}
.etl-table{border-collapse:collapse;width:100%;font-family:Arial,sans-serif;font-size:14px}
.etl-table th,.etl-table td{padding:8px 10px;border:1px solid #a7b0bd;text-align:right;vertical-align:top}
.etl-table th{font-weight:bold;background:#e8eef5;color:#152a43}
.etl-table td{white-space:pre-wrap;overflow-wrap:anywhere}
.etl-scroll{overflow-x:auto;margin:8px 0}
.etl-code{direction:ltr;text-align:left;white-space:pre-wrap;overflow-wrap:anywhere;font-family:Consolas,monospace;font-size:13px}
"""
display(HTML("<style>" + STYLE + "</style>"))

def rtl(text):
    display(HTML('<div class="etl-block" dir="rtl">' + html.escape(str(text)) + '</div>'))

def table_html(frame, title="", limit=15):
    frame = frame if isinstance(frame, pd.DataFrame) else pd.DataFrame(frame)
    shown = frame if limit is None else frame.head(limit)
    def value_html(value):
        if value is None or value is pd.NA:
            value = ""
        elif isinstance(value, (float, np.floating)):
            value = "" if np.isnan(value) else f"{value:,.4f}".rstrip("0").rstrip(".")
        return '<bdi dir="auto">' + html.escape(str(value)) + '</bdi>'
    head = ''.join('<th>' + value_html(c) + '</th>' for c in shown.columns)
    rows = ''.join('<tr>' + ''.join('<td>' + value_html(v) + '</td>' for v in row) + '</tr>'
                   for row in shown.itertuples(index=False, name=None))
    note = f'<p dir="rtl">מוצגות {len(shown)} שורות מתוך {len(frame)}.</p>' if len(shown) < len(frame) else ''
    return ('<section dir="rtl"><h3>' + html.escape(title) + '</h3><div class="etl-scroll">'
            '<table class="etl-table"><thead><tr>' + head + '</tr></thead><tbody>' + rows
            + '</tbody></table></div>' + note + '</section>')

def show_table(frame, title="", limit=15):
    display(HTML(table_html(frame, title, limit)))

show_table([{"מערכת": platform.system(), "Python": platform.python_version(),
             "Kernel": sys.executable, "פרויקט": str(PROJECT_ROOT)}], "הסביבה שנבחרה")
show_table(version_rows, "גרסאות החבילות", limit=None)
rtl("בדיקת הסביבה עברה. אפשר להמשיך לתיקיית העבודה.")
```

<h3>יצירת עותק עבודה להרצה הזאת</h3>
<p>כל הפקודות בהמשך יפעלו על העותק שייווצר בתא הבא. בתוך תיקיית ההרצה יהיו נתוני המקור, מסדי הנתונים, פלטי הבדיקות ותיקיית <code dir="ltr">exports</code>. הרצה נוספת של התא הזה יוצרת תיקיית הרצה חדשה; לאחר מכן מריצים שוב את כל התאים שמתחתיו.</p>

```python
RUN_ID = datetime.now().strftime("%Y%m%d_%H%M%S") + "_" + uuid.uuid4().hex[:4]
LAB_ROOT = PROJECT_ROOT / "notebook_lab" / RUN_ID
LAB_ROOT.mkdir(parents=True, exist_ok=False)
(LAB_ROOT / ".gitignore").write_text("*\n", encoding="utf-8")

base_files = [
    "requirements.txt", "requirements-core.txt", ".dlt/config.toml",
    "scripts/generate_data.py", "scripts/start_dagster.py", "scripts/warmup.py", "scripts/preflight.py",
    "pipelines/01_pure_python_etl.py", "pipelines/02_dlt_pipeline.py", "pipelines/03_rag_pipeline.py",
    "orchestration/definitions.py", "exercises/check_exercise.py",
    "dbt_project/dbt_project.yml", "dbt_project/profiles.yml",
    "dbt_project/macros/generate_schema_name.sql",
    "dbt_project/models/staging/_sources.yml", "dbt_project/models/staging/_staging.yml",
    "dbt_project/models/staging/stg_customers.sql", "dbt_project/models/staging/stg_orders.sql",
    "dbt_project/models/staging/stg_products.sql", "dbt_project/models/marts/_marts.yml",
    "dbt_project/models/marts/dim_customers.sql", "dbt_project/models/marts/fct_orders.sql",
    "dbt_project/models/marts/mart_daily_revenue.sql", "dbt_project/tests/assert_no_negative_revenue.sql",
]
base_files += [p.relative_to(PROJECT_ROOT).as_posix() for p in sorted((PROJECT_ROOT / "data/docs").glob("*.md"))]
base_files += [p.relative_to(PROJECT_ROOT).as_posix() for p in sorted((PROJECT_ROOT / "exercises/solution").iterdir())
               if p.is_file() and p.suffix in {".py", ".sql", ".yml"}]
source_hashes = {}
for relative in base_files:
    source, target = PROJECT_ROOT / relative, LAB_ROOT / relative
    target.parent.mkdir(parents=True, exist_ok=True)
    shutil.copy2(source, target)
    source_hashes[relative] = hashlib.sha256(source.read_bytes()).hexdigest()

LOG_DIR, EXPORT_DIR = LAB_ROOT / "logs", LAB_ROOT / "exports"
LOG_DIR.mkdir()
EXPORT_DIR.mkdir()
(LAB_ROOT / ".dagster_home").mkdir()
(LAB_ROOT / ".dagster_home/dagster.yaml").write_text("telemetry:\n  enabled: false\n", encoding="utf-8")
CHECKS, COMMANDS = {}, []
CHILD_ENV = os.environ.copy()
CHILD_ENV.update({
    "PATH": str(Path(sys.executable).parent) + os.pathsep + os.environ.get("PATH", ""),
    "PYTHONUTF8": "1", "PYTHONIOENCODING": "utf-8", "PYTHONUNBUFFERED": "1",
    "WAREHOUSE_PATH": str(LAB_ROOT / "warehouse.duckdb"),
    "DAGSTER_HOME": str(LAB_ROOT / ".dagster_home"),
    "GITHUB_OFFLINE": "1", "FAKE_EMBEDDINGS": "0", "DBT_USE_COLORS": "false",
    "DBT_SEND_ANONYMOUS_USAGE_STATS": "false", "HF_HUB_DISABLE_TELEMETRY": "1",
})

def check(condition, message):
    passed = bool(condition)
    CHECKS[message] = passed
    if not passed:
        raise AssertionError(message)
    rtl("עבר: " + message)

def run_cmd(arguments, label, cwd=None):
    started = time.perf_counter()
    completed = subprocess.run([str(a) for a in arguments], cwd=str(cwd or LAB_ROOT),
                               env=CHILD_ENV, capture_output=True, text=True,
                               encoding="utf-8", errors="replace", shell=False)
    elapsed = round(time.perf_counter() - started, 2)
    output = completed.stdout + ("\n" + completed.stderr if completed.stderr else "")
    log = LOG_DIR / f"{len(COMMANDS)+1:02d}_{label}.log"
    log.write_text(output, encoding="utf-8")
    COMMANDS.append({"stage": label, "exit_code": completed.returncode, "seconds": elapsed, "log": log.name})
    visible = re.sub(r"\x1b\[[0-?]*[ -/]*[@-~]", "", output)
    visible = visible.replace(str(LAB_ROOT), "<LAB>")
    if len(visible) > 24000:
        visible = visible[:3000] + "\n... full output in logs ...\n" + visible[-18000:]
    display(HTML('<details dir="rtl"><summary>פלט הרצה: ' + html.escape(label)
                 + f' | {elapsed} שניות | קוד יציאה {completed.returncode}</summary>'
                 + '<pre class="etl-code">' + html.escape(visible) + '</pre></details>'))
    if completed.returncode:
        raise RuntimeError(f"השלב {label} נכשל. פתח את הפלט שמעל או את {log}")
    return output

def run_script(relative, *arguments, label=None):
    return run_cmd([sys.executable, str(LAB_ROOT / relative), *arguments],
                   label or Path(relative).stem)

def query(sql, parameters=None):
    warehouse = LAB_ROOT / "warehouse.duckdb"
    if not warehouse.exists():
        raise RuntimeError("המחסן עדיין לא נוצר. הרץ תחילה את שלב dlt.")
    with closing(duckdb.connect(str(warehouse), read_only=True)) as connection:
        return connection.execute(sql, parameters or []).fetchdf()

def sqlite_query(sql):
    path = LAB_ROOT / "warehouse_basic.db"
    if not path.exists():
        raise RuntimeError("הרץ תחילה את שלב Python ETL.")
    with closing(sqlite3.connect(str(path))) as connection:
        return pd.read_sql_query(sql, connection)

def load_module(relative, short_name):
    name = "etl_lab_" + RUN_ID + "_" + short_name
    spec = importlib.util.spec_from_file_location(name, LAB_ROOT / relative)
    module = importlib.util.module_from_spec(spec)
    sys.modules[name] = module
    try:
        spec.loader.exec_module(module)
    except Exception:
        sys.modules.pop(name, None)
        raise
    return module

def build_dbt(label):
    executable = Path(sys.executable).parent / ("dbt.exe" if os.name == "nt" else "dbt")
    if not executable.exists():
        raise RuntimeError("dbt לא נמצא ליד ה־Python של ה־Kernel. בדוק את התקנת דרישות הפרויקט.")
    run_cmd([executable, "--no-send-anonymous-usage-stats", "build", "--profiles-dir", "."], label, LAB_ROOT / "dbt_project")
    result_path = LAB_ROOT / "dbt_project/target/run_results.json"
    payload = json.loads(result_path.read_text(encoding="utf-8"))
    (LOG_DIR / (label + "_run_results.json")).write_text(json.dumps(payload, ensure_ascii=False, indent=2), encoding="utf-8")
    results = pd.DataFrame([{"name": r["unique_id"], "status": r["status"],
                             "failures": r.get("failures")} for r in payload["results"]])
    show_table(results, "תוצאות dbt: מודלים ובדיקות", limit=None)
    check(results["status"].isin(["success", "pass", "warn"]).all(), label + ": כל המודלים והבדיקות הסתיימו ללא שגיאה")
    warnings = results[results["status"] == "warn"]
    check(len(warnings) == 1 and warnings["failures"].iloc[0] == 19,
          label + ": האזהרה הצפויה היא על 19 הפניות ללקוח חסר")
    return results

show_table([{"פריט": "תיקיית הריצה", "מיקום": str(LAB_ROOT)},
            {"פריט": "תוצאות לייצוא", "מיקום": str(EXPORT_DIR)},
            {"פריט": "פלטים מלאים", "מיקום": str(LOG_DIR)}], "היכן נשמרת ההרצה")
```
